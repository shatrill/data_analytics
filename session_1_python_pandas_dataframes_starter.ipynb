## Part 1: Meet Pandas

**Pandas** is Python's spreadsheet engine. It lets you load, clean, filter and summarize tables of data with a few lines of code, and it handles millions of rows that would crash Excel.

We load it with `import pandas as pd`. The `as pd` part is a nickname, so we can type `pd` instead of `pandas` every time.

````python
# Import the pandas library and give it the short nickname "pd"
# pandas = Python's toolkit for working with tables of data (called DataFrames)
import pandas as pd

print("pandas version:", pd.__version__)
````

### Load a real dataset from the internet
`pd.read_csv()` reads a CSV file (comma-separated values, the most common data file format) and turns it into a **DataFrame**, which is simply a table with rows and columns. Think of a DataFrame as one spreadsheet tab inside Python.

We will use the famous **Iris flowers** dataset: 150 flowers, each measured four ways, from three species.

````python
# Read the CSV straight from the web into a DataFrame named df
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/iris.csv"
df = pd.read_csv(url)

print("Data loaded!")
````

### Your three first-look commands

| Command | Plain-English meaning | Why analysts use it |
|---|---|---|
| `df.head()` | "Show me the first 5 rows." | A quick peek to see what the data looks like. |
| `df.info()` | "Tell me about each column." | Shows column names, data types (numbers vs. text), and how many values are missing. |
| `df.describe()` | "Give me the summary statistics." | Count, average (mean), spread (std), minimum, maximum and quartiles for every numeric column. |

Run each command in its own cell below.

````python
# 1) Peek at the first five rows
df.head(7)
````

````python
# 2) Column names, data types, and non-missing counts
df.info()
````

````python
# 3) Summary statistics for every numeric column
df.describe()
````

**What do you notice?** `info()` reports 150 non-null values in every column, so there is no missing data. `describe()` shows petal length ranges from about 1.0 cm to 6.9 cm, a big spread. We will find out why in Part 2.

## Part 2: Guided Example: Asking Questions of a DataFrame

Profiling tells us *what* is in the data. Now we start asking business-style questions. Follow along and run each step.

### Step 1: How big is the table?
`.shape` returns *(number of rows, number of columns)*. Note there are no parentheses after `shape`, because it is a property, not a command.

````python
print("Rows, Columns:", df.shape)
print("Column names:", list(df.columns))
````

### Step 2: Grab a single column
Use square brackets with the column name in quotes: `df["column_name"]`. A single column is called a **Series**.

````python
# Select the species column and show the first 5 values
df["species"].head(75)
````

### Step 3: Count the categories
`.value_counts()` counts how many times each value appears. This works like a quick pivot table.

````python
df["species"].value_counts()
````

### Step 4: Calculate a single statistic
You can call `.mean()`, `.median()`, `.min()`, `.max()` on any numeric column.

````python
avg_petal = df["petal_length"].mean()
print("Average petal length:", round(avg_petal, 2), "cm")
print("Longest petal:", df["petal_length"].max(), "cm")
print(avg_petal)
````

### Step 5: Filter rows with a condition
To keep only rows that meet a rule, put the condition inside `df[ ... ]`. Read it as: *"Give me the rows of df **where** petal length is greater than 5."* (This works just like SQL's `WHERE`.)

````python
big_petals = df[df["petal_length"] > 5]
#print(big_petals)
print("Flowers with petals over 5 cm:", len(big_petals))
big_petals["species"].value_counts()
````

### Step 6: Sort the table
`.sort_values()` sorts by a column. Add `ascending=False` to put the biggest values first.

````python
# The 5 flowers with the widest sepals
df.sort_values("sepal_width", ascending=False).head()
````

**💬 Insight:** Of the 42 flowers with petals over 5 cm, 41 are *virginica* and none are *setosa*. Petal length alone could almost identify a species. Finding patterns like this is exactly what analysts do.
