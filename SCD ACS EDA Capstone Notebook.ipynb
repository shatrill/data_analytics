## Step 1: Import Libraries

These libraries are commonly used in healthcare analytics.

```python
# Data manipulation
import pandas as pd
import numpy as np

# Visualization
import matplotlib.pyplot as plt
import seaborn as sns

# Display settings
pd.set_option("display.max_columns", None)
sns.set_theme(style="whitegrid")
```

## Step 2: Load the Dataset

Update the file path after uploading your CSV into Google Colab.

```python
# Example
file_path = "acs_sickle_cell_dataset.csv"

df = pd.read_csv(file_path)
print(df.shape)
df.head()
```

## Step 3: Understand the Data

We examine dimensions, data types, missing values, and summary statistics.

```python
# Overview of dataset
df.info()

# Missing values
display(df.isnull().sum().sort_values(ascending=False).head(20))

# Numeric summary
display(df.describe())
```

## Step 4: Mortality Analysis

```python
# Replace mortality column name if different
mortality_col = "mortality"

if mortality_col in df.columns:
    mortality_rate = df[mortality_col].mean()*100
    print(f"Mortality Rate: {mortality_rate:.2f}%")
```

## Step 5: Visualizations

```python
# Age distribution
plt.figure(figsize=(8,5))
sns.histplot(df.select_dtypes(include=np.number).iloc[:,0], bins=30)
plt.title("Distribution of First Numeric Variable")
plt.show()
```

```python
# Correlation heatmap
numeric_df = df.select_dtypes(include=np.number)

plt.figure(figsize=(12,8))
sns.heatmap(numeric_df.corr(), cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```
